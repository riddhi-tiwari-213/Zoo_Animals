```python
# Importing required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
# Dataset URL
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/zoo/zoo.data"

# Column names
columns = [
    'animal_name', 'hair', 'feathers', 'eggs', 'milk',
    'airborne', 'aquatic', 'predator', 'toothed',
    'backbone', 'breathes', 'venomous', 'fins',
    'legs', 'tail', 'domestic', 'catsize', 'class_type'
]

# Loading dataset
df = pd.read_csv(url, header=None, names=columns)

print("Dataset loaded successfully!")
```

```python
# Display first 5 records
print("First 5 rows:")
display(df.head())

# Display last 5 records
print("Last 5 rows:")
display(df.tail())

# Dataset shape
print("Dataset Shape:", df.shape)

# Dataset information
print("\nDataset Information:")
df.info()

# Statistical description
print("\nStatistical Description:")
display(df.describe())

# Description including categorical columns
print("\nComplete Dataset Description:")
display(df.describe(include='all'))
```

```python
# Mapping class numbers to animal types

animal_types = {
    1: 'Mammal',
    2: 'Bird',
    3: 'Reptile',
    4: 'Fish',
    5: 'Amphibian',
    6: 'Insect',
    7: 'Invertebrate'
}

df['animal_type'] = df['class_type'].map(animal_types)

# Count animals in each category
type_counts = df['animal_type'].value_counts()

print("Number of animals in each category:")
display(type_counts)

# Grouping using groupby()
print("\nGrouped animal counts:")
display(df.groupby('animal_type')['animal_name'].count().sort_values(ascending=False))
```

```python
# Counting animals having each characteristic

characteristics = ['hair', 'feathers', 'eggs', 'milk']

characteristic_counts = df[characteristics].sum()

print("Animals having each characteristic:")
display(characteristic_counts)

# Using NumPy for calculations
numpy_counts = np.array(characteristic_counts)

print("\nTotal counts using NumPy:")
print(numpy_counts)

# Finding the most common characteristic
most_common = characteristic_counts.idxmax()

print("\nMost common characteristic:", most_common)
print("Number of animals:", characteristic_counts.max())
```

```python
# Filtering animals which are both aquatic and predators

aquatic_predators = df[
    (df['aquatic'] == 1) & (df['predator'] == 1)
]

print("Animals that are both aquatic and predators:")

display(aquatic_predators[['animal_name', 'animal_type']])

print("Total number:", len(aquatic_predators))
```

```python
# Sorting animal categories by number of records

sorted_types = df['animal_type'].value_counts().sort_values(
    ascending=False
)

print("Animal types arranged in descending order:")
display(sorted_types)

# Category with maximum records
largest_type = sorted_types.idxmax()

print("\nAnimal type with the largest number of records:",
      largest_type)

print("Number of records:", sorted_types.max())
```

```python
# Checking missing values

print("Missing values in each column:")
display(df.isnull().sum())

print("\nTotal missing values:",
      df.isnull().sum().sum())

# Checking duplicate records

print("\nTotal duplicate rows:", df.duplicated().sum())

# Checking duplicate animal names

print("\nDuplicate animal names:",
      df['animal_name'].duplicated().sum())

# Displaying duplicate records, if any

duplicates = df[df.duplicated()]

if duplicates.empty:
    print("\nNo duplicate rows found.")
else:
    display(duplicates)
```

```python
# Converting selected characteristics into NumPy array

data_array = df[['hair', 'feathers', 'eggs',
                 'milk', 'aquatic', 'predator']].to_numpy()

print("NumPy Array:")
print(data_array)

# Total number of present characteristics
print("\nTotal present characteristics:",
      np.sum(data_array))

# Average number of selected characteristics per animal
print("\nAverage characteristics per animal:",
      np.mean(data_array))

# Maximum and minimum values
print("\nMaximum value:", np.max(data_array))
print("Minimum value:", np.min(data_array))
```

```python
# Bar chart showing animal categories

plt.figure(figsize=(9, 5))

type_counts.plot(kind='bar')

plt.title('Distribution of Animals by Type')
plt.xlabel('Animal Type')
plt.ylabel('Number of Animals')

plt.xticks(rotation=30)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Comparing selected animal characteristics

plt.figure(figsize=(8, 5))

characteristic_counts.plot(
    kind='bar',
    color='skyblue',
    edgecolor='black'
)

plt.title('Comparison of Animal Characteristics')
plt.xlabel('Characteristics')
plt.ylabel('Number of Animals')

plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
print("THREE OBSERVATIONS FROM ZOO ANIMALS DATASET")

# Observation 1
largest_type = type_counts.idxmax()
largest_count = type_counts.max()

print("\n1. Animal Category:")
print(f"{largest_type} has the largest number of records "
      f"with {largest_count} animals.")

# Observation 2
most_common = characteristic_counts.idxmax()
common_count = characteristic_counts.max()

print("\n2. Common Characteristic:")
print(f"{most_common} is the most common among the "
      f"four selected characteristics, appearing in "
      f"{common_count} animals.")

# Observation 3
aquatic_predator_count = len(aquatic_predators)

print("\n3. Aquatic Predators:")
print(f"{aquatic_predator_count} animals are both "
      f"aquatic and predators.")
```
